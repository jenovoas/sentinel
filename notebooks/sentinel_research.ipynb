````python
# Import required libraries
import sys
sys.path.append('..')  # Add parent directory to path

from sentinel_sdk import SentinelClient, SemanticVectors
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime, timedelta
import time

# Configure plotting
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")
%matplotlib inline

print("✅ Libraries imported successfully")
````

## 1. System Health Check

````python
# Initialize Sentinel client
client = SentinelClient("http://localhost:8000")

# Check if Sentinel is reachable
if client.ping():
    print("✅ Sentinel is online")
    
    # Get health status
    health = client.get_health()
    print(f"\nStatus: {health['status']}")
    print(f"Uptime: {health.get('uptime_seconds', 0):.1f}s")
    
    # Get semantic vectors
    vectors = client.get_semantic_vectors()
    print(f"\nSemantic Vectors:")
    print(f"  Coherence: {vectors.coherence:.2%}")
    print(f"  Entropy: {vectors.entropy:.4f}")
    print(f"  TTE: {vectors.tte_us:.2f}μs")
else:
    print("❌ Sentinel is not reachable. Please start it with 'sudo sctl start'")
````

## 2. Coherence Monitoring

Monitor system coherence over time to detect patterns and anomalies.

````python
# Collect coherence data
print("Collecting coherence data for 60 seconds...")

measurements = client.monitor_coherence(
    duration_seconds=60,
    interval=2.0,
    callback=lambda v: print(f"  {v.timestamp.strftime('%H:%M:%S')} - Coherence: {v.coherence:.4f}")
)

# Convert to DataFrame
df = pd.DataFrame([
    {
        'timestamp': m.timestamp,
        'coherence': m.coherence,
        'entropy': m.entropy,
        'tte_us': m.tte_us
    }
    for m in measurements
])

print(f"\n✅ Collected {len(df)} measurements")
df.head()
````

````python
# Statistical analysis
print("Statistical Analysis:")
print(f"\nCoherence:")
print(f"  Mean: {df['coherence'].mean():.4f}")
print(f"  Std: {df['coherence'].std():.4f}")
print(f"  Min: {df['coherence'].min():.4f}")
print(f"  Max: {df['coherence'].max():.4f}")

print(f"\nEntropy:")
print(f"  Mean: {df['entropy'].mean():.4f}")
print(f"  Std: {df['entropy'].std():.4f}")

print(f"\nTTE:")
print(f"  Mean: {df['tte_us'].mean():.2f}μs")
print(f"  Std: {df['tte_us'].std():.2f}μs")
````

````python
# Visualize coherence over time
fig, axes = plt.subplots(3, 1, figsize=(14, 10))

# Coherence
axes[0].plot(df['timestamp'], df['coherence'], linewidth=2, color='#00D9FF')
axes[0].axhline(y=0.90, color='red', linestyle='--', label='Threshold (0.90)', alpha=0.7)
axes[0].fill_between(df['timestamp'], df['coherence'], alpha=0.3, color='#00D9FF')
axes[0].set_ylabel('Coherence', fontsize=12, fontweight='bold')
axes[0].set_title('System Coherence Over Time', fontsize=14, fontweight='bold')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Entropy
axes[1].plot(df['timestamp'], df['entropy'], linewidth=2, color='#A855F7')
axes[1].fill_between(df['timestamp'], df['entropy'], alpha=0.3, color='#A855F7')
axes[1].set_ylabel('Entropy', fontsize=12, fontweight='bold')
axes[1].set_title('System Entropy Over Time', fontsize=14, fontweight='bold')
axes[1].grid(True, alpha=0.3)

# TTE
axes[2].plot(df['timestamp'], df['tte_us'], linewidth=2, color='#10B981')
axes[2].fill_between(df['timestamp'], df['tte_us'], alpha=0.3, color='#10B981')
axes[2].set_ylabel('TTE (μs)', fontsize=12, fontweight='bold')
axes[2].set_xlabel('Time', fontsize=12, fontweight='bold')
axes[2].set_title('Time-to-Equilibrium Over Time', fontsize=14, fontweight='bold')
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('coherence_analysis.png', dpi=300, bbox_inches='tight')
plt.show()

print("\n✅ Plot saved to coherence_analysis.png")
````

## 3. AI Performance Analysis

````python
# Test AI with different prompts
test_prompts = [
    "Explain quantum computing in simple terms",
    "What is machine learning?",
    "Describe the TCP/IP protocol",
    "How does encryption work?",
    "What is a neural network?"
]

ai_results = []

print("Testing AI performance...\n")
for prompt in test_prompts:
    response = client.query_ai(prompt, max_tokens=50)
    ai_results.append({
        'prompt': prompt[:40] + '...',
        'model': response.model,
        'latency_ms': response.latency_ms,
        'response_length': len(response.response)
    })
    print(f"✓ {prompt[:40]}... - {response.latency_ms:.2f}ms")

ai_df = pd.DataFrame(ai_results)
ai_df
````

````python
# Visualize AI latency
fig, ax = plt.subplots(figsize=(12, 6))

bars = ax.barh(range(len(ai_df)), ai_df['latency_ms'], color='#00D9FF', alpha=0.7)
ax.set_yticks(range(len(ai_df)))
ax.set_yticklabels(ai_df['prompt'])
ax.set_xlabel('Latency (ms)', fontsize=12, fontweight='bold')
ax.set_title('AI Query Latency by Prompt', fontsize=14, fontweight='bold')
ax.grid(True, alpha=0.3, axis='x')

# Add value labels
for i, bar in enumerate(bars):
    width = bar.get_width()
    ax.text(width, bar.get_y() + bar.get_height()/2, 
            f'{width:.1f}ms', ha='left', va='center', fontweight='bold')

plt.tight_layout()
plt.savefig('ai_latency_analysis.png', dpi=300, bbox_inches='tight')
plt.show()

print(f"\nAverage AI Latency: {ai_df['latency_ms'].mean():.2f}ms")
````

## 4. TruthSync Verification

````python
# Test scientific claims
scientific_claims = [
    "Water boils at 100°C at sea level",
    "The speed of light is approximately 299,792,458 m/s",
    "DNA is composed of four nucleotide bases",
    "The Earth is flat",  # False claim for testing
    "Gravity is a fundamental force"
]

verification_results = []

print("Verifying scientific claims...\n")
for claim in scientific_claims:
    result = client.verify_claim(claim)
    verification_results.append({
        'claim': claim[:50],
        'confidence': result.confidence,
        'status': result.status,
        'processing_us': result.processing_time_us
    })
    print(f"{'✓' if result.confidence > 0.7 else '✗'} {claim[:50]} - {result.confidence:.2%}")

verify_df = pd.DataFrame(verification_results)
verify_df
````

````python
# Visualize verification confidence
fig, ax = plt.subplots(figsize=(12, 6))

colors = ['#10B981' if c > 0.7 else '#EF4444' for c in verify_df['confidence']]
bars = ax.barh(range(len(verify_df)), verify_df['confidence'], color=colors, alpha=0.7)

ax.set_yticks(range(len(verify_df)))
ax.set_yticklabels(verify_df['claim'])
ax.set_xlabel('Confidence', fontsize=12, fontweight='bold')
ax.set_title('TruthSync Verification Confidence', fontsize=14, fontweight='bold')
ax.axvline(x=0.7, color='orange', linestyle='--', label='Threshold (0.7)', alpha=0.7)
ax.legend()
ax.grid(True, alpha=0.3, axis='x')

# Add value labels
for i, bar in enumerate(bars):
    width = bar.get_width()
    ax.text(width, bar.get_y() + bar.get_height()/2, 
            f'{width:.1%}', ha='left', va='center', fontweight='bold')

plt.tight_layout()
plt.savefig('truthsync_verification.png', dpi=300, bbox_inches='tight')
plt.show()
````

## 5. Correlation Analysis

````python
# Analyze correlations between metrics
correlation_matrix = df[['coherence', 'entropy', 'tte_us']].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0, 
            square=True, linewidths=1, cbar_kws={"shrink": 0.8})
plt.title('Correlation Matrix: Semantic Vectors', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('correlation_matrix.png', dpi=300, bbox_inches='tight')
plt.show()

print("\nCorrelation Analysis:")
print(correlation_matrix)
````
